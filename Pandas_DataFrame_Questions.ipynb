## Question 29:
Create a DataFrame named inventory_df from equal-length lists. Add a new column using a partially
indexed Series so that values appear only for selected row indexes. Display the result, then delete the added
column and display the DataFrame again.

```python
import pandas as pd

items = ['Laptop', 'Mouse', 'Keyboard', 'Monitor']
quantity = [15, 50, 45, 20]
price = [60000, 800, 1500, 12000]

inventory_df = pd.DataFrame({
    'Item': items,
    'Quantity': quantity,
    'Price': price
})

inventory_df['Discount'] = pd.Series([10, 15], index=[0, 2])
print(inventory_df)

inventory_df = inventory_df.drop(columns='Discount')
print(inventory_df)
```

## Question 30:
Copy a table with at least four columns from any webpage or spreadsheet and read it into lab_df. Write
one sequence of Pandas statements to: display the DataFrame, display its column names, retrieve one
single-word column, retrieve one column whose name contains a space, construct a smaller DataFrame from
selected columns, add a new whole column, and finally delete that column.

```python
import io
import pandas as pd

table_data = """EmpID	Employee Name	Department	Base Salary
101	Aarav Sharma	Engineering	85000
102	Diya Patel	Data Science	92000
103	Rohan Gupta	Marketing	64000
104	Ananya Verma	Finance	71000
"""

# Read copied data (use pd.read_clipboard() on local machine, or io.StringIO for Colab)
lab_df = pd.read_csv(io.StringIO(table_data), sep=r"\t+|\s\s+", engine="python")

# 1. Display the DataFrame
print(lab_df)

# 2. Display its column names
print(lab_df.columns)

# 3. Retrieve one single-word column
print(lab_df['Department'])

# 4. Retrieve one column whose name contains a space
print(lab_df['Employee Name'])

# 5. Construct a smaller DataFrame from selected columns
smaller_df = lab_df[['Employee Name', 'Base Salary']]
print(smaller_df)

# 6. Add a new whole column
lab_df['Status'] = 'Active'
print(lab_df)

# 7. Delete that column
lab_df = lab_df.drop(columns='Status')
print(lab_df)
```
